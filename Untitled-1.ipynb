```python
def read_dashboard_data(filename):
    dashboard = {}
    with open(filename, 'r') as file:
        next(file)
        for line in file:
            data = line.strip().split(';')
            print(data)
            if len(data) == 3:
                aiguillage_id = data[0]
                CS = int(data[1])
                addr = int(data[2])
                dashboard[aiguillage_id] = {'cs': CS, 'addr': addr}
    return dashboard


positions = read_dashboard_data('positions mapping.txt')
```

```python
toto = [int(item["cs"]) for i,item in positions.items()]
print(max(toto))
```
